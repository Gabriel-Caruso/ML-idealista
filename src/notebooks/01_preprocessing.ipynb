### OTRAS COLUMNAS CON CEROS

```python
# Aunque lo habíamos visto arriba, un bucle sencillo para ver proporción de ceros

numericas = ["PrecioActual", "PrecioAnterior", "metros", "habitaciones", "baños"]

for col in numericas:
    n_ceros = (df[col] == 0).sum()
    porcentaje = round((df[col] == 0).mean() * 100, 1)
    print(f"{col}: {n_ceros} ceros ({porcentaje}%)")
```

```python
# La que queda es precioAnterior, que tiene una barbaridad de ceros.
# Lo más probable es que no tuvieran datos anteriores sobre el precio de esa vivienda.

# Primero vamos a ver las que sí nos dan datos comparándolas con el precio actual
con_valor = df[df["PrecioAnterior"] > 0]

bajadas = (con_valor["PrecioAnterior"] > con_valor["PrecioActual"]).sum()
subidas = (con_valor["PrecioAnterior"] < con_valor["PrecioActual"]).sum()
iguales = (con_valor["PrecioAnterior"] == con_valor["PrecioActual"]).sum()

print("Filas con PrecioAnterior:", len(con_valor))
print("Bajadas de precio (anterior > actual):", bajadas)
print("Subidas (anterior < actual):", subidas)
print("iguales:", iguales)
```

### SOBRE ESTOS DATOS

Obviamente no son creíbles. El precio de la vivienda solo tiende a subir. Podemos sacar dos conclusiones respecto a esto: una mala y la otra peor.  

La mala: No hay datos fiables sobre el precio anterior de la vivienda y en este dataset se ha obtenido lo que se ha podido.  
La peor: Idealista instancia cada vivienda como un objeto nuevo cada vez que sube de precio. Sin embargo, no hace lo mismo con las bajadas de precio y por eso el 100% de los cambios de precio sobre LA MISMA VIVIENDA son de reducción. Son ofertas. Esto altera por completo las métricas y cualquier tipo de estudio. Así solo parece que las viviendas están estables y que, de vez en cuando, puedes encontrar ofertas.  

Sea como sea, lo que debemos de interpretar de esta columna entonces no es su literalidad, sino "si hay o no rebaja".

```python
# Vamos a entender que esta columna en realidad nos indica si hay o no rebaja.
df["flag_rebaja"] = (df["PrecioAnterior"] > 0).astype(int)
```

```python
# Y vamos a sacar otra columna para el porcentaje de rebaja. Igual es redundante, pero no está mal tenerla tampoco. Es una magnitud más que nos puede servir.
# Y además es un dato que se saca fila a fila. No hay imputación.
# TODO: Revisar si hay imputación por haber multiplicado por 100 y redondeado a 2.
diferencia = df["PrecioAnterior"] - df["PrecioActual"]

df["pct_rebaja"] = (np.where(df["PrecioAnterior"] > 0, diferencia / df["PrecioAnterior"], 0) * 100).round(2)
```

### COLUMNAS ASCENSOR Y LOCALIZACIÓN

```python
# Tiene 793 nulos.
# Para ver si son "nulos de verdad" lo mejor es cruzarlos contra el tipo de inmueble.
# Hipótesis: igual son Casas, no pisos, donde no puedes poner que "no tiene ascensor" porque obviamente no es que no tenga, sino que no puede existir

df.ascensor.value_counts(dropna=False)
```

```python
# Porcentaje de ascensor nulo dentro de cada tipo de inmueble
# Tengo que crear una columna porque no encuentro otra manera de analizar esto
# Después hago un drop de esta columna y listo.
# ESTO ES SOLO PARA EL ANÁLISIS, PARA CONFIRMAR LA HIPÓTESIS O DESCARTARLA
df["ascensor_es_nulo"] = df["ascensor"].isna()
porcentaje_nulos = (df.groupby("tipo_inmueble")["ascensor_es_nulo"].mean() * 100).round(2)
porcentaje_nulos
```

#### HIPÓTESIS CONFIRMADA:
Todas las casas o chalets tienen estos nulos.  
Algunos estudios (el 5.6%) tienen estos nulos.  
Muy pocos pisos (0.58%) tienen estos nulos.  

Lo suyo sería crear la columna de ascensor_limpio y estos nulos llamarlos NO_APLICA cuando son chalets o casas. Con los pisos se quedarían los nulos y hay dos posibilidades:  
Droppearlos si no son muchos (aunque un 5% es bastante)  
Meter un cuarto clasificador llamado DESCONOCIDO.

```python
df.tipo_inmueble.value_counts()
```

```python
chalet_casa = [ "Casa o chalet independiente", "Chalet adosado", "Chalet pareado", "Chalet", "Casa rural"]

es_chalet_casa = df["tipo_inmueble"].isin(chalet_casa)
es_chalet_casa
```

```python
# Sencillo: Creo la columna, compruebo si es True y además tiene el nulo en la columna, entonces "NO_APLICA"
# Después compruebo Si es false Y ADEMÁS tiene el nulo en la columna. Entonces "DESCONOCIDO"
df["ascensor_limpio"] = df["ascensor"]
df.loc[es_chalet_casa & df["ascensor"].isna(), "ascensor_limpio"] = "NO_APLICA"
df.loc[~es_chalet_casa & df["ascensor"].isna(), "ascensor_limpio"] = "DESCONOCIDO"  # Recordatorio de que en python/pandas la negación es con la virgulilla, no con exclamación
df.ascensor_limpio.value_counts(dropna=False)
```
